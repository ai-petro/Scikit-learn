```python
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# ========== 1. DATA GENERATION ==========
porosity = np.random.normal(0.2,0.05,100)
noise = np.random.normal(0,10,100)
w = 500
b = -50
permeability = w * porosity + b + noise 

# ========== 2. VISUALIZATION ==========
plt.scatter(porosity,permeability)
plt.gca()
plt.xlabel('porosity')
plt.ylabel('permeability')
plt.title('cross plot')

plt.show()

# ========== 3. MODEL TRAINING ==========
porosity_train  , porosity_test ,permeability_train, permeability_test = train_test_split(porosity,permeability,test_size=0.2) 

model = LinearRegression()
porosity_train = porosity_train.reshape(-1,1)
permeability_train = permeability_train.reshape(-1,1)
model.fit(porosity_train,permeability_train)

print(" W =" , model.coef_)
print(" B = ", model.intercept_)

# ========== 4. EVALUATION ==========
porosity_test = porosity_test.reshape(-1,1)
predict = model.predict(porosity_test)
print("Model predictions : ", predict)


mse = mean_squared_error(predict,permeability_test)
print(" MSE = " , mse)
rmse = np.sqrt(mse)
print(" RMSE = ", rmse)
```
